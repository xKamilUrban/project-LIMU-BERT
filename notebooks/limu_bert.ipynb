Tworzymy okna zawierajace po 120 probek, z nakladka 60

```python
import numpy as np
```

```python
STEP = 60
SEQ_LEN = 120

resapled_data = []
labels = []

all_windows = []
all_labels = []

for file in files:
  data = pd.read_csv(file)

  resapled_array = signal.resample_poly(data.iloc[:, :6], up=20, down=50, axis=0)
  labels = [data['label'][0]] * len(resapled_array)

  n_samples = len(resapled_array)

  for i in range(0, n_samples - SEQ_LEN + 1, STEP):
    window = resapled_array[i:i+120]
    if len(set(labels[i:i+120])) == 1:
      all_windows.append(window)
      all_labels.append(labels[0])


ACTIVITY_MAP = {
    'walking': 0,
    'sitting': 1,
    'standing': 2,
    'squats': 3,
    'pushups': 4,
    'jumping_jacks': 5
}

print()
all_labels = [ACTIVITY_MAP[act] for act in all_labels]

n = len(all_labels)
all_labels = [[[all_labels[i]]] * SEQ_LEN for i in range(n)]

X = np.array(all_windows, dtype=np.float32)
y = np.array(all_labels, dtype=np.float32)
print(X.shape)
print(y.shape)
```
